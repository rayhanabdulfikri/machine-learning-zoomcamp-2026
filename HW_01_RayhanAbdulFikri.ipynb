### **Q6. Median Value of Horsepower**
---
Find the median value of the horsepower column in the dataset.

1. Next, calculate the most frequent value of the same horsepower column.
2. Use the fillna method to fill the missing values in the horsepower column with the most frequent value from the previous step.
3. Now, calculate the median value of horsepower once again.

Has it changed?
* Yes, it increased
* Yes, it decreased
* No

```python
median_before = df['horsepower'].median()
print(median_before)
```

```python
mode_horsepower = df['horsepower'].mode()[0]
print(mode_horsepower)

# mode() returns a Series. Since there is only one mode value,
# I use [0] to extract that value.
```

```python
df['horsepower'] = df['horsepower'].fillna(mode_horsepower)
median_after = df['horsepower'].median()

print("Median before:", median_before)
print("Mode:", mode_horsepower)
print("Median after:", median_after)

if median_before < median_after:
  print("Yes, it increased")
elif median_before > median_after:
  print("Yes, it decreased")
else:
  print("No")
```

### **Q7. Median value of horsepower**
---
1. Select all the cars from Asia
2. Select only columns vehicle_weight and model_year
3. Select the first 7 values
4. Get the underlying NumPy array. Let's call it X.
5. Compute matrix-matrix multiplication between the transpose of X and X. To get the transpose, use X.T. Let's call the result XTX.
6. Invert XTX.
7. Create an array y with values [1100, 1300, 800, 900, 1000, 1100, 1200].
8. Multiply the inverse of XTX with the transpose of X, and then multiply the result by y. Call the result w.
9. What's the sum of all the elements of the result?


> Note: You just implemented linear regression. We'll talk about it in the next lesson.



```python
# 1. Select all the cars from Asia
asia = df[df["origin"] == "Asia"]

# 2. Select only columns vehicle_weight and model_year
X = asia[["vehicle_weight", "model_year"]]

# 3. Select the first 7 values
X = X.head(7)

# 4. Get the underlying NumPy array. Let's call it X
X = X.values

# 5. Compute matrix-matrix multiplication between X.T and X
XTX = X.T @ X

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create an array y
y = np.array([
    1100,
    1300,
    800,
    900,
    1000,
    1100,
    1200
])

# 8. Calculate w
w = XTX_inv @ X.T @ y

# 9. Find the sum of all the elements of w
print("w =", w)
print("sum(w) =", round(w.sum(), 3))

import plotly.graph_objects as go

# 10. Create a grid for the regression plane
vehicle_weight = np.linspace(
    X[:, 0].min(),
    X[:, 0].max(),
    30
)

model_year = np.linspace(
    X[:, 1].min(),
    X[:, 1].max(),
    30
)

W, Y = np.meshgrid(
    vehicle_weight,
    model_year
)

# 11. Calculate the regression plane
Z = w[0] * W + w[1] * Y

# 12. Create the interactive regression plane
fig = go.Figure()

fig.add_trace(
    go.Scatter3d(
        x=X[:, 0],
        y=X[:, 1],
        z=y,
        mode="markers",
        marker=dict(size=7),
        name="Actual Data",
        text=[
            f"Vehicle Weight: {X[i, 0]}<br>"
            f"Model Year: {X[i, 1]}<br>"
            f"y: {y[i]}"
            for i in range(len(X))
        ],
        hovertemplate="%{text}<extra></extra>"
    )
)

fig.add_trace(
    go.Surface(
        x=W,
        y=Y,
        z=Z,
        opacity=0.5,
        name="Regression Plane"
    )
)

# 13. Configure the chart
fig.update_layout(
    title="Interactive Multiple Linear Regression",
    scene=dict(
        xaxis_title="Vehicle Weight",
        yaxis_title="Model Year",
        zaxis_title="y"
    ),
    width=1000,
    height=700
)

# 14. Show the interactive visualization
fig.show()
```
